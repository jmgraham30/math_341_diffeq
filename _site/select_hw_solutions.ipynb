# Homework Solutions

This page links to full solutions for select homework exercises for the
Fall 2026 semester. Each solution is worked out “by hand” and then,
where helpful, confirmed symbolically or numerically using Python.
Solutions are posted after the corresponding homework has been
collected, and this page may be updated during the semester.

These solutions do not replace the course textbook (Logan, *A First
Course in Differential Equations*, 3rd ed.) or lectures.

- [Homework 1](solutions_01.qmd): Section 1.1 (verifying exponential
  solutions)
- [Homework 2](solutions_02.qmd): Slope fields, separable equations,
  first-order linear equations and integrating factors, and RC circuits
- [Homework 3](solutions_03.qmd): Autonomous equations, phase lines, and
  bifurcations
- [Homework 4](solutions_04.qmd): Numerical output, second-order linear
  homogeneous equations, damping and oscillation, undetermined
  coefficients, and variation of parameters
- [Homework 5](solutions_05.qmd): Laplace transforms and inverse
  transforms (Sections 3.1 and 3.2)
- [Homework 6](solutions_06.qmd): Laplace transforms and initial value
  problems (Section 3.2)
- [Homework 7](solutions_07.qmd): Convolution, impulsive sources, and
  $2\times 2$ matrices and linear systems
