# Homework Solutions

This page links to full solutions for select homework exercises for the
Fall 2026 semester. Each solution is worked out “by hand” and then,
where helpful, confirmed symbolically or numerically using Python.
Solutions are posted after the corresponding homework has been
collected, and this page may be updated during the semester.

These solutions do not replace the course textbook (Logan, *A First
Course in Differential Equations*, 3rd ed.) or lectures.

- [Homework 1](https://math341diffeq.netlify.app/solutions_01): Section
  1.1 (verifying exponential solutions)

- [Homework 2](https://math341diffeq.netlify.app/solutions_02): Slope
  fields, separable equations, first-order linear equations and
  integrating factors, and RC circuits

- [Homework 3](https://math341diffeq.netlify.app/solutions_03):
  Autonomous equations, phase lines, and bifurcations

- [Homework 4](https://math341diffeq.netlify.app/solutions_04):
  Numerical output, second-order linear homogeneous equations, damping
  and oscillation, undetermined coefficients, and variation of
  parameters

- [Homework 5](https://math341diffeq.netlify.app/solutions_05): Laplace
  transforms and inverse transforms (Sections 3.1 and 3.2)

- [Homework 6](https://math341diffeq.netlify.app/solutions_06): Laplace
  transforms and initial value problems (Section 3.2)

- [Homework 7](https://math341diffeq.netlify.app/solutions_07):
  Convolution, impulsive sources, and $2\times 2$ matrices and linear
  systems
