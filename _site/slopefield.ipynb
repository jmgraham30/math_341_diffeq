# Slope Field Explorer

Direction fields, solution curves, and equilibria for first-order
equations x’ = f(t, x)

Enter a first-order equation $x' = f(t,x)$ or choose an example. Click
anywhere in the plane to draw the solution through that point (traced
forward and backward in $t$), or type an initial condition $x(t_0)=x_0$.
Use the parameter sliders to explore how the slope field changes, show
the curve $f=0$ and other isoclines, and, for autonomous equations, see
the equilibrium solutions classified as stable, unstable, or
semi-stable. The Euler’s method overlay connects this page to the
Numerical Methods Explorer.
