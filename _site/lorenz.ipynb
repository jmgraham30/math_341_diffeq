# Lorenz Attractor Explorer

Chaos, sensitive dependence, and the geometry of a strange attractor

The Lorenz system
$$x' = \sigma(y-x),\qquad y' = x(\rho - z) - y,\qquad z' = xy - \beta z$$
is a three-dimensional autonomous system whose solutions, for
$\sigma = 10$, $\beta = 8/3$, and $\rho = 28$, neither settle to an
equilibrium nor to a periodic orbit. Drag the picture to rotate it,
slide $\rho$ to watch the behavior change, and use the tabs below the
picture to compare two nearby solutions (sensitive dependence), view the
time series, and see the Lorenz map of successive maxima of $z$.
