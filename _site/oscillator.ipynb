# Oscillator and RLC Circuit Explorer

Damped, driven, and switched second-order systems: one equation, three
views

A mass on a spring and a series RLC circuit obey the same equation,
$$m x'' + b x' + k x = F(t) \qquad\text{and}\qquad L Q'' + R Q' + \tfrac{1}{C} Q = E(t),$$
which in the form used in the notes is
$x'' + 2\gamma x' + \omega_0^2 x = f(t)$ with $\gamma = b/2m$,
$\omega_0^2 = k/m$, and $f = F/m$. Change the parameters, the initial
condition, or the input, and watch the same solution in three views: the
graph of $x(t)$, the **roots** of the characteristic equation in the
complex plane, and the **phase portrait**. The panel on the right shows
the steady-state amplitude and phase for a cosine drive. Use the “RLC
circuit” button to switch the labels and the picture to the electrical
version.

## Guided experiments

Each experiment starts with a prediction. Write down your answer
*before* you open the link, then check it.

### 1. The three damping regimes (Notes 4)

Take $m = 1$, $k = 4$, so $\omega_0 = 2$, and start from $x(0)=1$,
$x'(0)=0$.

- **Predict:** as $b$ increases from $0$, where do the two roots move?
  At what value of $b$ do they meet, and what happens to the solution
  when they do?
- Check with <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=0.4&amp;k=4&amp;x0=1&amp;v0=0&amp;in=none"
  target="_blank"><span class="math inline"><em>b</em> = 0.4</span>
  (underdamped)</a>, <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=4&amp;k=4&amp;x0=1&amp;v0=0&amp;in=none"
  target="_blank"><span class="math inline"><em>b</em> = 4</span>
  (critically damped)</a>, and <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=5&amp;k=4&amp;x0=1&amp;v0=0&amp;in=none"
  target="_blank"><span class="math inline"><em>b</em> = 5</span>
  (overdamped)</a>. The teal curve shows the path the roots take as $b$
  increases.
- **Explain:** why is the overdamped solution slower to settle than the
  critically damped one, even though it has more damping? (Look at the
  slow root.)

### 2. Resonance without damping (Examples 7 and 8)

- **Predict:** for $x'' + 4x = \cos(\omega t)$ starting from rest, what

  target="_blank"><span class="math inline"><em>ω</em> = 1</span></a>
  (the solution is $(\cos t - \cos 2t)/3$), <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=0&amp;k=4&amp;x0=0&amp;v0=0&amp;in=cos&amp;F0=1&amp;w=2"
  target="_blank"><span class="math inline"><em>ω</em> = 2</span></a>
  (the solution is $t\sin(2t)/4$, which grows linearly), and <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=0&amp;k=4&amp;x0=0&amp;v0=0&amp;in=cos&amp;F0=1&amp;w=1.8&amp;T=40"
  target="_blank"><span class="math inline"><em>ω</em> = 1.8</span></a>

- **Explain:** where is the point $i\omega$ in the root plot compared

### 3. Resonance with damping: transient and steady state

- **Predict:** with $b = 0.4$, will the largest amplitude occur exactly
  at $\omega = \omega_0$? Will it be the same for $b = 2$?
- Open <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=0.4&amp;k=4&amp;x0=0&amp;v0=0&amp;in=cos&amp;F0=1&amp;w=2"
  target="_blank">the damped resonance setup</a>, tick “Transient” to
  see $x - x_p$, and use the buttons $\omega=\omega_0$ and
  $\omega=\omega_r$. The orange dot on the right shows $D(\omega)$ and
  the phase $\delta$.
- **Explain:** the transient dies like $e^{-\gamma t}$, but the
  steady-state oscillation does not. Why can the steady state be larger
  than the static deflection $f/\omega_0^2$ when
  $\omega \approx \omega_0$? Compare with $b = 2$, where the peak
  formula $\omega_r = \sqrt{\omega_0^2 - 2\gamma^2}$ gives a smaller
  value and there is only a mild peak.

### 4. Switching on a constant force (Laplace unit)

- **Predict:** a force $F_0 = 4$ is switched on at $t=0$ for $m=1$,
  $k=4$. Where will the mass finally rest? For $b = 0.8$, will it
  overshoot, and by how much?
- Check with <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=0.8&amp;k=4&amp;x0=0&amp;v0=0&amp;in=step&amp;F0=4"
  target="_blank"><span class="math inline"><em>b</em> = 0.8</span></a>
  and <a
  href="apps/oscillator.html#mode=mech&amp;m=1&amp;b=4&amp;k=4&amp;x0=0&amp;v0=0&amp;in=step&amp;F0=4"
  target="_blank"><span class="math inline"><em>b</em> = 4</span></a>.
  The Facts panel reports the overshoot, which for a start from rest
  should match $e^{-\pi\gamma/\omega_d}$.
- **Explain:** how do the roots in the left half-plane predict the
  overshoot and the settling time?

- Check with <a
