## Reference: Key Laplace Transform Facts

The **Laplace transform** of $f(t)$, defined for $t \geq 0$, is $$
F(s) = \mathcal{L}[f(t)] = \int_0^{\infty} e^{-st} f(t)\,dt,
$$ whenever the integral converges.

### Short Table

The following entries are used repeatedly in the examples below.

| $f(t)$                     | $F(s) = \mathcal{L}[f(t)]$ | Conditions |
|----------------------------|----------------------------|------------|
| $1$                        | $\dfrac{1}{s}$             | $s > 0$    |
| $t^n$ ($n = 0,1,2,\ldots$) | $\dfrac{n!}{s^{n+1}}$      | $s > 0$    |
| $e^{at}$                   | $\dfrac{1}{s-a}$           | $s > a$    |
| $\sin(kt)$                 | $\dfrac{k}{s^2+k^2}$       | $s > 0$    |
| $\cos(kt)$                 | $\dfrac{s}{s^2+k^2}$       | $s > 0$    |
| $\sinh(kt)$                | $\dfrac{k}{s^2-k^2}$       | $s > |k|$  |
| $\cosh(kt)$                | $\dfrac{s}{s^2-k^2}$       | $s > |k|$  |
| $t^n e^{at}$               | $\dfrac{n!}{(s-a)^{n+1}}$  | $s > a$    |
| $e^{at}\sin(kt)$           | $\dfrac{k}{(s-a)^2+k^2}$   | $s > a$    |
| $e^{at}\cos(kt)$           | $\dfrac{s-a}{(s-a)^2+k^2}$ | $s > a$    |
| $H(t-a)\,f(t-a)$           | $e^{-as}F(s)$              | $s > 0$    |

### Operational Properties

**Linearity.** $$
\mathcal{L}[\alpha f(t) + \beta g(t)] = \alpha F(s) + \beta G(s).
$$

**First Shift (Exponential Shift) Property.** If
$\mathcal{L}[f(t)] = F(s)$, then $$
\mathcal{L}[e^{at}f(t)] = F(s - a).
$$

**Second Shift (Switching) Property.** If $\mathcal{L}[f(t)] = F(s)$,
then $$
\mathcal{L}[H(t-a)\,f(t-a)] = e^{-as}F(s),
$$ or equivalently in inverse form, $$
\mathcal{L}^{-1}[e^{-as}F(s)] = H(t-a)\,f(t-a).
$$

## Part I — Forward Transforms (Exercise 8)

### Example 1 — Linearity and the Table: $6 + 5e^{-2t} + te^{3t}$

> Find $\mathcal{L}\!\left[6 + 5e^{-2t} + te^{3t}\right]$.

#### By Hand

Apply linearity and read each piece from the table.

**Term 1:**
$\mathcal{L}[6] = \mathcal{L}[6 \cdot 1] = 6 \cdot \dfrac{1}{s} = \dfrac{6}{s}$.

**Term 2:**
$\mathcal{L}[5e^{-2t}] = 5 \cdot \dfrac{1}{s-(-2)} = \dfrac{5}{s+2}$.

**Term 3:** $\mathcal{L}[te^{3t}]$. The table entry for $t^n e^{at}$
with $n=1$, $a=3$ gives $\dfrac{1!}{(s-3)^2} = \dfrac{1}{(s-3)^2}$.

Combining by linearity: $$
\boxed{\mathcal{L}\!\left[6 + 5e^{-2t} + te^{3t}\right]
= \frac{6}{s} + \frac{5}{s+2} + \frac{1}{(s-3)^2}.}
$$

#### Using SymPy

````python
f1 = 6 + 5*sym.exp(-2*t) + t*sym.exp(3*t)

F1 = sym.laplace_transform(f1, t, s, noconds=True)
display(Math(r'\mathcal{L}\bigl[6 + 5e^{-2t} + te^{3t}\bigr] = '
            + sym.latex(sym.simplify(F1))))
````

------------------------------------------------------------------------

### Example 2 — Switching Property: $tH(t-3)$

> Find $\mathcal{L}[tH(t-3)]$.

#### By Hand

The function $tH(t-3)$ switches on at $t = 3$, but it is not yet in the
standard switching form $H(t-a)f(t-a)$ because the function being
switched is $f(t) = t$, not $f(t-3)$. We rewrite $t$ in terms of
$(t-3)$: $$
t = (t - 3) + 3.
$$ Therefore $$
tH(t-3) = (t-3)H(t-3) + 3H(t-3).
$$

Each piece is now in standard switching form. Apply the switching
property $\mathcal{L}[H(t-a)f(t-a)] = e^{-as}F(s)$ with $a = 3$:

- $\mathcal{L}[(t-3)H(t-3)]$: here $f(t) = t$, so
  $F(s) = \dfrac{1}{s^2}$, giving $e^{-3s}\cdot\dfrac{1}{s^2}$.
- $\mathcal{L}[3H(t-3)]$: here $f(t) = 3$, so $F(s) = \dfrac{3}{s}$,
  giving $e^{-3s}\cdot\dfrac{3}{s}$.

Combining: $$
\boxed{\mathcal{L}[tH(t-3)] = e^{-3s}\!\left(\frac{1}{s^2} + \frac{3}{s}\right)
= \frac{e^{-3s}(1 + 3s)}{s^2}.}
$$

> **Tip**
>
> **The rewriting trick.** Whenever you see $f(t)H(t-a)$ and $f(t)$ is
> not naturally expressed as a function of $(t-a)$, write $t = (t-a)+a$
> (or expand $f(t)$ about $t=a$) to put the expression into standard
> switching form.

#### Using SymPy

````python
f2 = t * sym.Heaviside(t - 3)

F2 = sym.laplace_transform(f2, t, s, noconds=True)
display(Math(r'\mathcal{L}[t\,H(t-3)] = ' + sym.latex(sym.simplify(F2))))
````

### Example 3 — Direct Table Look-up: $\cos 5t$

> Find $\mathcal{L}[\cos 5t]$.

#### By Hand

From the table with $k = 5$: $$
\mathcal{L}[\cos(5t)] = \frac{s}{s^2 + 25}.
$$

$$\boxed{\mathcal{L}[\cos 5t] = \frac{s}{s^2+25}.}$$

#### Using SymPy

````python
F3 = sym.laplace_transform(sym.cos(5*t), t, s, noconds=True)
display(Math(r'\mathcal{L}[\cos 5t] = ' + sym.latex(F3)))
````

### Example 4 — Angle Addition: $\sin(2t + \pi)$

> Find $\mathcal{L}[\sin(2t + \pi)]$.

#### By Hand

The function $\sin(2t + \pi)$ is not directly in the table, but we can
expand it using the angle addition formula: $$
\sin(2t + \pi) = \sin(2t)\cos\pi + \cos(2t)\sin\pi = -\sin(2t) + 0 = -\sin(2t).
$$

Now apply linearity and the table entry for $\sin(kt)$ with $k = 2$: $$
\mathcal{L}[\sin(2t+\pi)] = \mathcal{L}[-\sin(2t)] = -\frac{2}{s^2+4}.
$$

$$\boxed{\mathcal{L}[\sin(2t+\pi)] = -\frac{2}{s^2+4}.}$$

> **Note**
>
> This result can also be understood geometrically: $\sin(2t + \pi)$ is
> a half-period phase shift of $\sin(2t)$, which simply negates the
> function.

#### Using SymPy

````python
f4 = sym.sin(2*t + sym.pi)

F4 = sym.laplace_transform(f4, t, s, noconds=True)
display(Math(r'\mathcal{L}[\sin(2t+\pi)] = ' + sym.latex(sym.simplify(F4))))
````

### Example 5 — Shift Property with Hyperbolic Function: $3e^{-t}\cosh t$

> Find $\mathcal{L}[3e^{-t}\cosh t]$.

#### By Hand

We use the **first shift property**: $\mathcal{L}[e^{at}f(t)] = F(s-a)$,
where $F(s) = \mathcal{L}[f(t)]$.

Here $a = -1$ and $f(t) = 3\cosh t$.

From the table, $\mathcal{L}[\cosh(kt)] = \dfrac{s}{s^2 - k^2}$. With
$k = 1$: $$
\mathcal{L}[\cosh t] = \frac{s}{s^2-1},
$$ so $\mathcal{L}[3\cosh t] = \dfrac{3s}{s^2-1}$.

By the shift property (replacing $s$ with $s - (-1) = s + 1$): $$
\mathcal{L}[3e^{-t}\cosh t] = \frac{3(s+1)}{(s+1)^2 - 1}.
$$

Expand the denominator: $$
(s+1)^2 - 1 = s^2 + 2s + 1 - 1 = s^2 + 2s = s(s+2).
$$

$$\boxed{\mathcal{L}[3e^{-t}\cosh t] = \frac{3(s+1)}{s(s+2)}.}$$

#### Using SymPy

````python
f5 = 3*sym.exp(-t)*sym.cosh(t)

F5 = sym.laplace_transform(f5, t, s, noconds=True)
display(Math(r'\mathcal{L}[3e^{-t}\cosh t] = '
            + sym.latex(sym.simplify(F5))))
````

------------------------------------------------------------------------

### Example 6 — Switching Property: $H(t-\pi)\cos(t-\pi)$

> Find $\mathcal{L}[H(t-\pi)\cos(t-\pi)]$.

#### By Hand

This is already in the standard switching form $H(t-a)f(t-a)$ with
$a = \pi$ and $f(t) = \cos t$.

From the table: $F(s) = \mathcal{L}[\cos t] = \dfrac{s}{s^2+1}$.

By the switching property: $$
\mathcal{L}[H(t-\pi)\cos(t-\pi)] = e^{-\pi s} \cdot \frac{s}{s^2+1}.
$$

$$\boxed{\mathcal{L}[H(t-\pi)\cos(t-\pi)] = \frac{se^{-\pi s}}{s^2+1}.}$$

#### Using SymPy

````python
f6 = sym.Heaviside(t - sym.pi) * sym.cos(t - sym.pi)

F6 = sym.laplace_transform(f6, t, s, noconds=True)
display(Math(r'\mathcal{L}[H(t-\pi)\cos(t-\pi)] = '
            + sym.latex(sym.simplify(F6))))
````

## Part II — Inverse Transforms (Exercise 9)

The principal techniques for finding inverse Laplace transforms are:

1.  **Direct table look-up** — recognize $F(s)$ immediately.
2.  **Linearity** — split sums and pull out constants.
3.  **Partial fraction decomposition** — break up rational functions.
4.  **Completing the square** — handle shifted quadratic denominators.
5.  **Switching property** — handle factors of $e^{-as}$.

------------------------------------------------------------------------

### Example 7 — Direct Table: $\dfrac{7}{s+2}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{7}{s+2}\right]$.

#### By Hand

Recognise $\dfrac{1}{s-a}$ with $a = -2$, scaled by $7$: $$
\mathcal{L}^{-1}\!\left[\frac{7}{s+2}\right]
= 7\,\mathcal{L}^{-1}\!\left[\frac{1}{s-(-2)}\right]
= 7e^{-2t}.
$$

$$\boxed{\mathcal{L}^{-1}\!\left[\frac{7}{s+2}\right] = 7e^{-2t}.}$$

#### Using SymPy

````python
F7 = sym.Rational(7,1) / (s + 2)

f7 = sym.inverse_laplace_transform(F7, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{7}{s+2}\right] = '
            + sym.latex(f7)))
````

------------------------------------------------------------------------

### Example 8 — Linearity and Table: $\dfrac{3}{s} - \dfrac{2}{s^2+6}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{3}{s} - \dfrac{2}{s^2+6}\right]$.

#### By Hand

Split using linearity.

**Term 1:**
$\mathcal{L}^{-1}\!\left[\dfrac{3}{s}\right] = 3 \cdot 1 = 3$.

**Term 2:** $\dfrac{2}{s^2+6}$. The table form for $\sin(kt)$ is
$\dfrac{k}{s^2+k^2}$, so we need $k = \sqrt{6}$ and must manufacture
that factor in the numerator: $$
\frac{2}{s^2+6}
= \frac{2}{\sqrt{6}} \cdot \frac{\sqrt{6}}{s^2+6}.
$$

Therefore $\mathcal{L}^{-1}\!\left[\dfrac{2}{s^2+6}\right]
= \dfrac{2}{\sqrt{6}}\sin(\sqrt{6}\,t)$.

Combining: $$
\boxed{\mathcal{L}^{-1}\!\left[\frac{3}{s} - \frac{2}{s^2+6}\right]
= 3 - \frac{2}{\sqrt{6}}\sin(\sqrt{6}\,t).}
$$

> **Tip**
>
> **Adjusting the numerator.** When the denominator is $s^2 + k^2$ but
> the numerator is a constant $c \neq k$, write
> $\dfrac{c}{s^2+k^2} = \dfrac{c}{k}\cdot\dfrac{k}{s^2+k^2}$ and invert
> to get $\dfrac{c}{k}\sin(kt)$.

#### Using SymPy

````python
F8 = sym.Rational(3,1)/s - sym.Rational(2,1)/(s**2 + 6)

f8 = sym.inverse_laplace_transform(F8, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{3}{s}-\frac{2}{s^2+6}\right] = '
            + sym.latex(sym.simplify(f8))))
````

------------------------------------------------------------------------

### Example 9 — Shift Property: $\dfrac{2}{(s-5)^2}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{2}{(s-5)^2}\right]$.

#### By Hand

Recognise the shifted form $\dfrac{n!}{(s-a)^{n+1}}$ with $n = 1$ and
$a = 5$: $$
\frac{2}{(s-5)^2} = 2 \cdot \frac{1!}{(s-5)^{1+1}}.
$$

From the table,
$\mathcal{L}^{-1}\!\left[\dfrac{1!}{(s-5)^2}\right] = te^{5t}$, so: $$
\boxed{\mathcal{L}^{-1}\!\left[\frac{2}{(s-5)^2}\right] = 2te^{5t}.}
$$

#### Using SymPy

````python
F9 = sym.Integer(2) / (s - 5)**2

f9 = sym.inverse_laplace_transform(F9, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{2}{(s-5)^2}\right] = '
            + sym.latex(f9)))
````

------------------------------------------------------------------------

### Example 10 — Switching Property: $\dfrac{7}{s}e^{-4s}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{7}{s}e^{-4s}\right]$.

#### By Hand

The factor $e^{-4s}$ signals the switching property with $a = 4$. The
remaining factor is $F(s) = \dfrac{7}{s}$, whose inverse transform is
$f(t) = 7$.

By the switching property
$\mathcal{L}^{-1}[e^{-as}F(s)] = H(t-a)f(t-a)$: $$
\mathcal{L}^{-1}\!\left[\frac{7}{s}e^{-4s}\right]
= 7H(t-4).
$$

$$\boxed{\mathcal{L}^{-1}\!\left[\frac{7}{s}e^{-4s}\right] = 7H(t-4).}$$

This represents a step function that turns on (jumps to the value $7$)
at $t = 4$.

#### Using SymPy

````python
F10 = (sym.Integer(7)/s) * sym.exp(-4*s)

f10 = sym.inverse_laplace_transform(F10, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{7}{s}e^{-4s}\right] = '
            + sym.latex(f10)))
````

### Example 11 — Partial Fractions with Switching: $\dfrac{1}{s(s-2)}e^{-s}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{1}{s(s-2)}e^{-s}\right]$.

#### By Hand

**Step 1 — Factor out $e^{-s}$.**

The factor $e^{-s}$ signals the switching property with $a = 1$. We
first find $\mathcal{L}^{-1}\!\left[\dfrac{1}{s(s-2)}\right]$ and then
apply the switching property.

**Step 2 — Partial fraction decomposition of $\dfrac{1}{s(s-2)}$.**

$$
\frac{1}{s(s-2)} = \frac{A}{s} + \frac{B}{s-2}.
$$

Multiply both sides by $s(s-2)$: $$
1 = A(s-2) + Bs.
$$

Set $s = 0$: $1 = A(-2) \Rightarrow A = -\dfrac{1}{2}$.

Set $s = 2$: $1 = B(2) \Rightarrow B = \dfrac{1}{2}$.

So $$
\frac{1}{s(s-2)} = -\frac{1/2}{s} + \frac{1/2}{s-2}.
$$

**Step 3 — Invert the partial fractions.**

$$
\mathcal{L}^{-1}\!\left[\frac{1}{s(s-2)}\right]
= -\frac{1}{2}\cdot 1 + \frac{1}{2}e^{2t}
= \frac{1}{2}\!\left(e^{2t} - 1\right) =: f(t).
$$

**Step 4 — Apply the switching property with $a = 1$.**

$$
\mathcal{L}^{-1}\!\left[\frac{e^{-s}}{s(s-2)}\right]
= H(t-1)\,f(t-1)
= H(t-1)\cdot\frac{1}{2}\!\left(e^{2(t-1)} - 1\right).
$$

$$\boxed{\mathcal{L}^{-1}\!\left[\frac{e^{-s}}{s(s-2)}\right]
= \frac{1}{2}H(t-1)\!\left(e^{2(t-1)}-1\right).}$$

#### Using SymPy

````python
F11 = sym.exp(-s) / (s*(s - 2))

f11 = sym.inverse_laplace_transform(F11, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{e^{-s}}{s(s-2)}\right] = '
            + sym.latex(sym.simplify(f11))))
````

### Example 12 — Mixed Sinusoidal: $\dfrac{7s+1}{s^2+4}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{7s+1}{s^2+4}\right]$.

#### By Hand

Split the fraction into a $\cos$ part and a $\sin$ part using the table
forms $\dfrac{s}{s^2+k^2}$ and $\dfrac{k}{s^2+k^2}$ with $k = 2$:

$$
\frac{7s+1}{s^2+4}
= 7\cdot\frac{s}{s^2+4} + \frac{1}{s^2+4}.
$$

For the second term, introduce the required factor of $k = 2$ in the
numerator: $$
\frac{1}{s^2+4} = \frac{1}{2}\cdot\frac{2}{s^2+4}.
$$

Therefore: $$
\mathcal{L}^{-1}\!\left[\frac{7s+1}{s^2+4}\right]
= 7\cos(2t) + \frac{1}{2}\sin(2t).
$$

$$\boxed{\mathcal{L}^{-1}\!\left[\frac{7s+1}{s^2+4}\right]
= 7\cos(2t) + \frac{1}{2}\sin(2t).}$$

#### Using SymPy

````python
F12 = (7*s + 1) / (s**2 + 4)

f12 = sym.inverse_laplace_transform(F12, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{7s+1}{s^2+4}\right] = '
            + sym.latex(sym.simplify(f12))))
````

### Example 13 — Rescaling the Frequency: $\dfrac{3}{2s^2+7}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{3}{2s^2+7}\right]$.

#### By Hand

The table form requires the denominator to be $s^2 + k^2$. Factor out
$2$: $$
\frac{3}{2s^2+7} = \frac{3}{2}\cdot\frac{1}{s^2+7/2}.
$$

This matches $\dfrac{k}{s^2+k^2}$ with $k^2 = \dfrac{7}{2}$, i.e.  
$k = \sqrt{7/2} = \dfrac{\sqrt{14}}{2}$. Write: $$
\frac{3}{2(s^2+7/2)}
= \frac{3}{2} \cdot \frac{1}{\sqrt{7/2}} \cdot \frac{\sqrt{7/2}}{s^2+7/2}
= \frac{3}{2} \cdot \sqrt{\frac{2}{7}} \cdot \frac{\sqrt{7/2}}{s^2+7/2}
= \frac{3}{\sqrt{14}} \cdot \frac{\sqrt{7/2}}{s^2+7/2}.
$$

Therefore: $$
\mathcal{L}^{-1}\!\left[\frac{3}{2s^2+7}\right]
= \frac{3}{\sqrt{14}}\sin\!\left(\frac{\sqrt{14}}{2}\,t\right).
$$

$$\boxed{\mathcal{L}^{-1}\!\left[\frac{3}{2s^2+7}\right]
= \frac{3}{\sqrt{14}}\sin\!\left(\!\sqrt{\frac{7}{2}}\,t\right).}$$

#### Using SymPy

````python
F13 = sym.Integer(3) / (2*s**2 + 7)

f13 = sym.inverse_laplace_transform(F13, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{3}{2s^2+7}\right] = '
            + sym.latex(sym.simplify(f13))))
````

### Example 14 — Switching with Cosine: $e^{-\pi s}\dfrac{3s}{s^2+9}$

> Find $\mathcal{L}^{-1}\!\left[e^{-\pi s}\dfrac{3s}{s^2+9}\right]$.

#### By Hand

The factor $e^{-\pi s}$ signals the switching property with $a = \pi$.
The remaining factor $F(s) = \dfrac{3s}{s^2+9}$ matches the table entry
for $\cos(kt)$ with $k = 3$: $$
\mathcal{L}^{-1}\!\left[\frac{3s}{s^2+9}\right]
= 3\,\mathcal{L}^{-1}\!\left[\frac{s}{s^2+9}\right]
= 3\cos(3t) =: f(t).
$$

Applying the switching property: $$
\mathcal{L}^{-1}\!\left[e^{-\pi s}\frac{3s}{s^2+9}\right]
= H(t-\pi)\,f(t-\pi)
= 3H(t-\pi)\cos(3(t-\pi)).
$$

$$\boxed{\mathcal{L}^{-1}\!\left[e^{-\pi s}\frac{3s}{s^2+9}\right]
= 3H(t-\pi)\cos(3(t-\pi)).}$$

> **Note**
>
> Note that
> $\cos(3(t-\pi)) = \cos(3t - 3\pi) = \cos(3t)\cos(3\pi) + \sin(3t)\sin(3\pi) = -\cos(3t)$,
> so the solution can also be written as $-3H(t-\pi)\cos(3t)$, though
> the $(t-\pi)$ form makes the switching structure clearer.

#### Using SymPy

````python
F14 = sym.exp(-sym.pi*s) * (3*s / (s**2 + 9))

f14 = sym.inverse_laplace_transform(F14, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[e^{-\pi s}\frac{3s}{s^2+9}\right] = '
            + sym.latex(sym.simplify(f14))))
````

### Example 15 — Completing the Square: $\dfrac{5s}{(s-3)^2+4}$

> Find $\mathcal{L}^{-1}\!\left[\dfrac{5s}{(s-3)^2+4}\right]$.

#### By Hand

The denominator $(s-3)^2 + 4$ is already in completed-square form with
$a = 3$ and $k = 2$. The table forms involving this denominator are: $$
\mathcal{L}^{-1}\!\left[\frac{s-a}{(s-a)^2+k^2}\right] = e^{at}\cos(kt),
\qquad
\mathcal{L}^{-1}\!\left[\frac{k}{(s-a)^2+k^2}\right] = e^{at}\sin(kt).
$$

We need to write the numerator $5s$ in terms of $(s-3)$: $$
5s = 5(s-3) + 15.
$$

Therefore: $$
\frac{5s}{(s-3)^2+4}
= \frac{5(s-3) + 15}{(s-3)^2+4}
= 5\cdot\frac{s-3}{(s-3)^2+4} + 15\cdot\frac{1}{(s-3)^2+4}.
$$

For the second term, introduce the factor $k = 2$ in the numerator: $$
15\cdot\frac{1}{(s-3)^2+4}
= \frac{15}{2}\cdot\frac{2}{(s-3)^2+4}.
$$

Now invert term by term using $a = 3$, $k = 2$: $$
\mathcal{L}^{-1}\!\left[\frac{5s}{(s-3)^2+4}\right]
= 5e^{3t}\cos(2t) + \frac{15}{2}e^{3t}\sin(2t).
$$

$$\boxed{\mathcal{L}^{-1}\!\left[\frac{5s}{(s-3)^2+4}\right]
= e^{3t}\!\left[5\cos(2t) + \frac{15}{2}\sin(2t)\right].}$$

#### Using SymPy

````python
F15 = 5*s / ((s - 3)**2 + 4)

f15 = sym.inverse_laplace_transform(F15, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[\frac{5s}{(s-3)^2+4}\right] = '
            + sym.latex(sym.simplify(f15))))
````

### Example 16 — Switching with a Ramp: $e^{-2s}\dfrac{3}{s^2}$

> Find $\mathcal{L}^{-1}\!\left[e^{-2s}\dfrac{3}{s^2}\right]$.

#### By Hand

The factor $e^{-2s}$ signals the switching property with $a = 2$. The
remaining factor is $F(s) = \dfrac{3}{s^2}$. Using the table entry
$\mathcal{L}[t^n] = \dfrac{n!}{s^{n+1}}$ with $n = 1$: $$
\mathcal{L}^{-1}\!\left[\frac{3}{s^2}\right]
= 3\,\mathcal{L}^{-1}\!\left[\frac{1!}{s^2}\right]
= 3t =: f(t).
$$

Applying the switching property: $$
\mathcal{L}^{-1}\!\left[e^{-2s}\frac{3}{s^2}\right]
= H(t-2)\,f(t-2)
= 3(t-2)H(t-2).
$$

$$\boxed{\mathcal{L}^{-1}\!\left[e^{-2s}\frac{3}{s^2}\right]
= 3(t-2)H(t-2).}$$

This is a **ramp** that begins at $t = 2$ and grows with slope $3$
thereafter.

#### Using SymPy

````python
F16 = sym.exp(-2*s) * sym.Integer(3) / s**2

f16 = sym.inverse_laplace_transform(F16, s, t)
display(Math(r'\mathcal{L}^{-1}\!\left[e^{-2s}\frac{3}{s^2}\right] = '
            + sym.latex(sym.simplify(f16))))
````

````python
t_vals = np.linspace(0, 5, 500)
f_delayed = 3*(t_vals - 2)*np.where(t_vals >= 2, 1, 0)
f_undelayed = 3*t_vals

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(t_vals, f_delayed,   color='tomato',    lw=2.2,
        label=r'$3(t-2)H(t-2)$')
ax.plot(t_vals, f_undelayed, color='steelblue', lw=1.4, ls='--',
        label=r'$3t$ (undelayed)')
ax.axvline(2, color='gray', lw=1, ls=':')
ax.annotate(r'$t=2$', xy=(2, 0.2), xytext=(2.1, 0.5), fontsize=10, color='gray')
ax.set_xlabel(r'$t$', fontsize=13)
ax.set_ylabel(r'$f(t)$', fontsize=13)
ax.set_title(r'Delayed ramp: $\mathcal{L}^{-1}\!\left[e^{-2s}\,3/s^2\right]$',
             fontsize=13)
ax.legend(fontsize=10)
ax.set_ylim(-0.5, 10)
plt.tight_layout()
plt.show()
````

------------------------------------------------------------------------

> **Summary: Strategy for Inverse Laplace Transforms**
>
> Work through these steps in order:
>
> 1.  **Look for $e^{-as}$.** If present, factor it out, find
>     $\mathcal{L}^{-1}[F(s)]$ of the remaining piece, then apply the
>     switching property:
>     $\mathcal{L}^{-1}[e^{-as}F(s)] = H(t-a)f(t-a)$.
>
> 2.  **Check for a direct table match.** $\dfrac{c}{s-a}$,
>     $\dfrac{c}{s^2+k^2}$, $\dfrac{cs}{s^2+k^2}$,
>     $\dfrac{c}{(s-a)^{n+1}}$ all have immediate inverses.
>
> 3.  **Split mixed numerators.** If the numerator is $\alpha s + \beta$
>     and the denominator is $s^2 + k^2$ (or $(s-a)^2 + k^2$), split
>     into separate $\cos$/$\sin$ (or $e^{at}\cos$/$e^{at}\sin$)
>     fractions.
>
> 4.  **Partial fractions.** For rational $F(s)$ with distinct or
>     repeated real roots in the denominator, decompose into simpler
>     fractions before inverting.
>
> 5.  **Complete the square.** If the denominator is a quadratic
>     $s^2 + bs + c$ with complex roots, write it as $(s-a)^2 + k^2$ and
>     use the shifted table entries.
