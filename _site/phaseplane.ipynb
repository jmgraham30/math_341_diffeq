# Phase Plane Explorer

An interactive replacement for pplane

Type any autonomous system $x' = f(x,y),\ y' = g(x,y)$, or pick an
example, then **click in the plane** to start a trajectory. Nullclines,
equilibria, and their linearization-based classification update as you
change the system or the parameters $a,b,c,d$. For a linear system
$\mathbf{x}' = A\mathbf{x}$ the matrix is
$A = \begin{pmatrix} a & b\\ c & d\end{pmatrix}$.
