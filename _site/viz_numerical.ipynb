# Numerical Methods Explorer

Euler, modified Euler (RK2), and RK4 for first-order initial value
problems

Choose an equation $x' = f(t,x)$, then adjust the step size $h$ and the
initial condition to compare Euler’s method, the modified Euler (RK2)
method, and RK4 against the exact solution. Use the tabs to view
solution curves, the log–log convergence plot, a step-by-step table, and
the formulas.
